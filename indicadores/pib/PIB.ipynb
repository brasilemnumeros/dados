# PIB — variação real anual

Este notebook é o motor de coleta e publicação da série `pib-real-anual`.

- **Fonte:** IBGE, Contas Nacionais — variação em volume do PIB a preços de mercado. O
  número é lido na série **SGS 7326** do Banco Central, que republica a taxa do IBGE com 2
  casas decimais.
- **Endpoint:** `https://api.bcb.gov.br/dados/serie/bcdata.sgs.7326/dados?formato=json`
- **Definição:** variação percentual do PIB real (volume) do ano civil em relação ao ano
  civil anterior. Unidade `percentual` (`%`, 2 casas decimais), `eixo: percentual`,
  `medida: variacao-anual`, granularidade `anual`. A data é o **início** do ano de
  referência (`AAAA-01-01`): o ponto de `2015-01-01` é a variação de 2015 contra 2014.
- **Fórmula:** nenhuma. A taxa é definida pela fonte; o notebook busca a série, recorta os
  anos fechados, confere contra o IBGE e formata.
- **Período publicado:** `1995-01-01` a `2025-01-01` (31 pontos).
- **Arquivo gerado:** `indicadores/pib/pib-real-anual.json`, com `versao: 1` e
  `id: pib-real-anual`.

## Decisão metodológica

O IBGE **publica** a taxa pronta em duas tabelas, ambas com **1 casa decimal**: a **6784**
(Contas Nacionais Anuais, variável 9810 — “PIB – variação em volume”), de 1996 a **2023**, e a
**5932** (Contas Nacionais Trimestrais, variável 6563 — “Taxa acumulada ao longo do ano”), cujo
4º trimestre é o ano fechado, de 1996 a **2025**.

A série publicada fica na **SGS 7326** por três motivos: (a) entrega a mesma taxa com **2
casas**, a precisão que a série já publica; (b) cobre até **2025**, enquanto a tabela anual para
em 2023; (c) a SGS 7326 é idêntica, ao centésimo, à taxa recomposta a partir do **índice de
volume encadeado** do IBGE (SIDRA 1620, “PIB a preços de mercado”, base média 1995 = 100) em
todos os anos de 1997 a 2025 — identidade conferida a cada execução, contra a taxa oficial
(tolerância de 0,05 p.p., o arredondamento de 1 casa) e contra o índice (0,01 p.p.).

A tabela anual 6784 **não** é usada como fonte porque, nesta edição, ela discorda das tabelas
de nível em **2018 e 2019** (1,7 e 1,3 contra 1,8 e 1,2 em 1620/5932, e 1,78/1,22 na SGS). A
execução imprime essa divergência; a escolha é pelas tabelas de nível, que concordam entre si.

## Recorte de início e ano fechado

A série começa em **1995**, alinhada às demais famílias do repositório, que começam em
`1995-01-01`. A SGS 7326 vai até 1963, mas os anos anteriores a 1995 vêm da série antiga (base
de volume anterior): a cadeia atual começa em 1995 — o índice do IBGE tem base média 1995 = 100
e o primeiro trimestre publicado é 1996T1. Estender ganharia 32 pontos (1963–1994) ao custo de
misturar bases de volume, por isso o corte é mantido.

O ponto de **1995** é o único que as tabelas atuais do IBGE não permitem conferir: a cadeia de
volume começa no nível de 1995 e a primeira taxa publicada é a de 1996. Ele vem da SGS 7326
(4,22%) e é mantido para preservar o início de 1995 das demais séries; quem quiser estritamente
a base atual pode cortar a série em 1996.

O ano corrente nunca é publicado: em novembro de 2026 a fonte já fecha **2024** (3,42%) e
**2025** (2,29%), e 2026 ainda não tem taxa anual.

## Coleta da série oficial

A leitura é feita desde o início da série, sem atualização incremental e sem ler o JSON já
publicado para decidir o período. A SGS devolve uma leitura por ano, datada em `01/01`.


````python
from datetime import date
from decimal import Decimal, ROUND_HALF_UP
from pathlib import Path
import json
import math
import os
import time

import requests

RAIZ = Path(os.environ.get("DADOS_RAIZ", Path.cwd())).resolve()
PASTA = RAIZ / "indicadores" / "pib" if (RAIZ / "indicadores" / "pib").is_dir() else RAIZ
DESTINO = Path(os.environ.get("PIB_SAIDA", PASTA))
ARQUIVO = DESTINO / "pib-real-anual.json"

SERIE = 7326
URL_SGS = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.{serie}/dados"
URL_FONTE = URL_SGS.format(serie=SERIE) + "?formato=json"
PRIMEIRO_ANO = 1995
HOJE = date.today()


def baixar_json(url, params=None, tentativas=4):
    """GET com espera crescente; SGS e SIDRA falham de forma intermitente."""
    for tentativa in range(tentativas):
        try:
            resposta = requests.get(url, params=params, timeout=120)
            resposta.raise_for_status()
            return resposta.json()
        except (requests.RequestException, ValueError):
            if tentativa == tentativas - 1:
                raise
            time.sleep(2 ** tentativa)


bruto = baixar_json(URL_SGS.format(serie=SERIE), {"formato": "json"})
if not bruto:
    raise ValueError(f"a série SGS {SERIE} não retornou dados")

anual = sorted(
    (date(int(item["data"][-4:]), int(item["data"][3:5]), int(item["data"][:2])), float(item["valor"]))
    for item in bruto
)

if len({dia for dia, _ in anual}) != len(anual):
    raise ValueError(f"a SGS {SERIE} devolveu anos repetidos")
if any((dia.day, dia.month) != (1, 1) for dia, _ in anual):
    raise ValueError(f"a SGS {SERIE} devolveu data fora de 01/01")
for (anterior, _), (seguinte, _) in zip(anual, anual[1:]):
    if seguinte.year - anterior.year != 1:
        raise ValueError(f"lacuna na SGS {SERIE}: {anterior} -> {seguinte}")

print(f"Série SGS {SERIE}: {len(anual)} leituras anuais")
print(f"Período bruto: {anual[0][0]:%Y-%m-%d} a {anual[-1][0]:%Y-%m-%d}")
print(f"Arquivo de saída: {ARQUIVO}")
print("\nPrimeiras 3 leituras:")
for dia, valor in anual[:3]:
    print(f"  {dia:%Y-%m-%d}  {valor:7.2f}%")
print("Últimas 3 leituras:")
for dia, valor in anual[-3:]:
    print(f"  {dia:%Y-%m-%d}  {valor:7.2f}%")
````

## Recorte dos anos fechados e conferência contra o IBGE

O recorte começa em 1995 e exclui o ano corrente, que ainda não tem taxa anual. Depois, cada
ponto é conferido contra as duas publicações do IBGE: a taxa oficial das Contas Nacionais
Trimestrais (SIDRA 5932, variável 6563, 4º trimestre) e o índice de volume encadeado (SIDRA
1620, variável 583, “PIB a preços de mercado”), recompondo o ano pela razão entre a soma dos 4
trimestres do ano e a do ano anterior. Divergência acima da tolerância interrompe a publicação.

A tabela anual 6784 é lida apenas para registro: ela para em 2023 e, nesta edição, discorda das
tabelas de nível em 2018 e 2019.


````python
URL_SIDRA = (
    "https://apisidra.ibge.gov.br/values/t/{tabela}/n1/all/v/{variavel}/p/all"
    "/c11255/{categoria}?formato=json"
)
PIB_A_PRECOS_DE_MERCADO = "90707"
SEM_DADO = {"..", "...", "-", "X"}


def sidra_trimestral(tabela, variavel):
    """Leituras trimestrais do PIB a preços de mercado: {\"202504\": valor}."""
    linhas = baixar_json(
        URL_SIDRA.format(tabela=tabela, variavel=variavel, categoria=PIB_A_PRECOS_DE_MERCADO)
    )
    return {
        linha["D3C"]: float(linha["V"])
        for linha in linhas[1:]
        if linha["V"] not in SEM_DADO
    }


fechados = [
    (dia, valor) for dia, valor in anual if PRIMEIRO_ANO <= dia.year < HOJE.year
]
if not fechados:
    raise ValueError(f"nenhum ano fechado a partir de {PRIMEIRO_ANO}")
if fechados[-1][0].year < HOJE.year - 1:
    print(f"AVISO: a fonte não tem o ano {HOJE.year - 1}; último ano fechado é {fechados[-1][0].year}")

# Taxa oficial do IBGE: "Taxa acumulada ao longo do ano" (6563); no 4º trimestre é o ano civil.
taxa_ibge = {
    int(periodo[:4]): valor
    for periodo, valor in sidra_trimestral(5932, 6563).items()
    if periodo.endswith("4")
}

# Índice de volume encadeado (base média 1995 = 100), por ano e trimestre.
indice_ibge = {}
for periodo, valor in sidra_trimestral(1620, 583).items():
    indice_ibge.setdefault(int(periodo[:4]), {})[int(periodo[4:])] = valor

# A taxa anual é a razão entre a soma dos 4 trimestres do ano e a do ano anterior, menos 1.
composta_ibge = {}
for ano in sorted(indice_ibge):
    atual = indice_ibge[ano]
    anterior = indice_ibge.get(ano - 1, {})
    if len(atual) == 4 and len(anterior) == 4:
        composta_ibge[ano] = (sum(atual.values()) / sum(anterior.values()) - 1) * 100

TOLERANCIA_TAXA = 0.051    # meio décimo: o IBGE publica a taxa com 1 casa decimal
TOLERANCIA_INDICE = 0.01   # o índice é publicado com 2 casas decimais

# A série do IBGE começa em 1996: a cadeia de volume tem nível em 1995, mas não taxa.
SEM_CONFERENCIA_IBGE = {1995}

for dia, valor in fechados:
    ano = dia.year
    if ano in SEM_CONFERENCIA_IBGE:
        continue
    if ano not in taxa_ibge:
        raise ValueError(f"{ano} sem taxa no SIDRA 5932")
    if abs(valor - taxa_ibge[ano]) > TOLERANCIA_TAXA:
        raise ValueError(f"{ano}: SGS {valor} x taxa IBGE {taxa_ibge[ano]}")
    # O índice só permite recompor a partir de 1997 (falta 1995 para a base de 1996).
    if ano in composta_ibge and abs(valor - composta_ibge[ano]) > TOLERANCIA_INDICE:
        raise ValueError(f"{ano}: SGS {valor} x índice IBGE {composta_ibge[ano]:.4f}")

print(f"Anos publicados: {len(fechados)} ({fechados[0][0]:%Y} a {fechados[-1][0]:%Y})")
print(
    f"Conferidos contra a taxa do IBGE (SIDRA 5932/6563): "
    f"{len(fechados) - len(SEM_CONFERENCIA_IBGE)} anos; "
    f"1995 fica fora porque o IBGE só publica taxa a partir de 1996."
)
print(
    f"Conferidos contra o índice de volume (SIDRA 1620/583): "
    f"{len(fechados) - len(SEM_CONFERENCIA_IBGE) - 1} anos; "
    f"1996 fica fora porque o índice começa em 1996T1 e a base de 1996 é 1995."
)
print("\nAno  SGS 7326  taxa IBGE  índice IBGE")
for dia, valor in fechados:
    if dia.year not in (1995, 1996, 2003, 2009, 2015, 2016, 2020, 2021, 2024, 2025):
        continue
    taxa = taxa_ibge.get(dia.year)
    composta = composta_ibge.get(dia.year)
    texto_taxa = "—" if taxa is None else f"{taxa:.1f}"
    texto_indice = "—" if composta is None else f"{composta:.3f}"
    print(f"{dia.year}  {valor:8.2f}  {texto_taxa:>9}  {texto_indice:>11}")

# Registro: a tabela anual (6784, variável 9810) não é a fonte e discorda das tabelas de nível.
linhas_anuais = baixar_json(
    "https://apisidra.ibge.gov.br/values/t/6784/n1/all/v/9810/p/all?formato=json"
)
anual_ibge = {
    int(linha["D3C"]): float(linha["V"])
    for linha in linhas_anuais[1:]
    if linha["V"] not in SEM_DADO
}
divergentes = [
    (dia.year, valor, anual_ibge[dia.year])
    for dia, valor in fechados
    if dia.year in anual_ibge and abs(valor - anual_ibge[dia.year]) > TOLERANCIA_TAXA
]
print(
    f"\nTabela 6784 (Contas Nacionais Anuais): "
    f"{min(anual_ibge)}–{max(anual_ibge)}, {len(anual_ibge)} anos."
)
if divergentes:
    detalhe = ", ".join(f"{ano} ({sgs:.2f} x {ibge:.1f})" for ano, sgs, ibge in divergentes)
    print(f"  divergências acima de {TOLERANCIA_TAXA} p.p.: {detalhe}")
    print("  não usada como fonte: as tabelas de nível (1620) e a taxa trimestral (5932) concordam com a SGS.")
````

## Escrita do JSON do contrato

O arquivo é gravado inteiro (nada de merge com o que já está publicado), com `versao`, `id`,
`unidade`, `eixo`, `medida`, `granularidade`, `fonte`, `urlFonte` e `dados` no formato
`{"data": "YYYY-MM-DD", "valor": ...}`.


````python
def arredondar(valor):
    """Arredonda para 2 casas com meio para cima (convenção comercial)."""
    return float(Decimal(repr(valor)).quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))


pontos = [
    {"data": dia.strftime("%Y-%m-%d"), "valor": arredondar(valor)} for dia, valor in fechados
]

serie = {
    "versao": 1,
    "id": "pib-real-anual",
    "unidade": {"tipo": "percentual", "simbolo": "%", "casas": 2},
    "eixo": "percentual",
    "medida": "variacao-anual",
    "granularidade": "anual",
    "fonte": "IBGE",
    "urlFonte": URL_FONTE,
    "dados": pontos,
}

DESTINO.mkdir(parents=True, exist_ok=True)
with ARQUIVO.open("w", encoding="utf-8") as arquivo:
    json.dump(serie, arquivo, ensure_ascii=False, indent=2)
    arquivo.write("\n")

print(f"{len(pontos)} pontos escritos em {ARQUIVO}")
print(f"Primeiro ponto: {pontos[0]}")
print(f"Último ponto:   {pontos[-1]}")
````

## Validação inline

A célula final falha se o contrato, a ordem, a unicidade dos anos, a finitude dos valores, a
igualdade com a fonte ou a conferência contra o IBGE forem violados. Também confere os números
citados em `docs/metodologia.md`.


````python
CHAVES_RAIZ = {
    "versao", "id", "unidade", "eixo", "medida", "granularidade", "fonte", "urlFonte", "dados",
}
assert set(serie) == CHAVES_RAIZ, sorted(set(serie) ^ CHAVES_RAIZ)
assert serie["versao"] == 1, serie["versao"]
assert serie["id"] == "pib-real-anual", serie["id"]
assert serie["unidade"] == {"tipo": "percentual", "simbolo": "%", "casas": 2}, serie["unidade"]
assert serie["eixo"] == "percentual", serie["eixo"]
assert serie["medida"] == "variacao-anual", serie["medida"]
assert serie["granularidade"] == "anual", serie["granularidade"]
assert serie["fonte"] == "IBGE", serie["fonte"]
assert serie["urlFonte"].startswith("https://"), serie["urlFonte"]

dados = serie["dados"]
assert dados, "série vazia"

datas = [ponto["data"] for ponto in dados]
dias = [date.fromisoformat(texto) for texto in datas]
por_ano = {ponto["data"][:4]: ponto["valor"] for ponto in dados}

for ponto in dados:
    assert set(ponto) == {"data", "valor"}, sorted(ponto)
    assert isinstance(ponto["data"], str), ponto
    assert not isinstance(ponto["valor"], bool), ponto
    assert isinstance(ponto["valor"], (int, float)), ponto
    assert math.isfinite(ponto["valor"]), ponto

assert all(a < b for a, b in zip(dias, dias[1:])), "datas fora de ordem ou repetidas"
assert all((dia.day, dia.month) == (1, 1) for dia in dias), "data fora de 01/01"
assert all(b.year - a.year == 1 for a, b in zip(dias, dias[1:])), "ano faltando ou repetido"
assert datas[0] == "1995-01-01", datas[0]
assert datas[-1] == "2025-01-01", datas[-1]
assert len(dados) == len(fechados), (len(dados), len(fechados))

# 1) O valor publicado é a leitura da fonte, arredondada a 2 casas.
esperado = [
    {"data": dia.strftime("%Y-%m-%d"), "valor": arredondar(valor)} for dia, valor in fechados
]
assert dados == esperado, "valor publicado difere da fonte"

# 2) Cada ponto continua coerente com o IBGE (taxa oficial e índice de volume).
for dia, valor in fechados:
    if dia.year in SEM_CONFERENCIA_IBGE:
        continue
    assert abs(valor - taxa_ibge[dia.year]) <= TOLERANCIA_TAXA, (dia.year, valor, taxa_ibge[dia.year])
    if dia.year in composta_ibge:
        assert abs(valor - composta_ibge[dia.year]) <= TOLERANCIA_INDICE, (dia.year, valor)

# 3) Números citados em docs/metodologia.md (esta edição da fonte).
citados = {
    "1995": 4.22,
    "2003": 1.14,
    "2009": -0.13,
    "2015": -3.55,
    "2016": -3.28,
    "2020": -3.28,
    "2021": 4.76,
    "2024": 3.42,
    "2025": 2.29,
}
for ano, valor in citados.items():
    assert por_ano[ano] == valor, (ano, por_ano[ano], valor)

# 4) O PIB de 2020 encolheu e o de 2021 recuperou: o sinal tem de refletir o choque.
assert por_ano["2020"] < 0 < por_ano["2021"], (por_ano["2020"], por_ano["2021"])

with ARQUIVO.open(encoding="utf-8") as arquivo:
    salvo = json.load(arquivo)
assert salvo == serie, "arquivo gravado difere do objeto validado"

print(
    f"ok: {len(dados)} anos únicos, {datas[0]} a {datas[-1]}, "
    f"leituras idênticas à SGS {SERIE} e coerentes com o SIDRA (5932 e 1620)"
)
````
