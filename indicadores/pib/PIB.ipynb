# PIB — variação real e per capita

Este notebook é o motor de coleta e publicação de três séries da família `pib`:
`pib-real-anual`, `pib-per-capita-nominal` e `pib-per-capita-real`.

## Variação real anual (`pib-real-anual`)

- **Fonte:** IBGE, Contas Nacionais — variação em volume do PIB a preços de mercado. O
  número é lido na série **SGS 7326** do Banco Central, que republica a taxa do IBGE com 2
  casas decimais.
- **Endpoint:** `https://api.bcb.gov.br/dados/serie/bcdata.sgs.7326/dados?formato=json`
- **Definição:** variação percentual do PIB real (volume) do ano civil em relação ao ano
  civil anterior. Unidade `percentual` (`%`, 2 casas decimais), `eixo: percentual`,
  `medida: variacao-anual`, granularidade `anual`. A data é o **início** do ano de
  referência (`AAAA-01-01`): o ponto de `2015-01-01` é a variação de 2015 contra 2014.
- **Fórmula:** nenhuma. A taxa é definida pela fonte; o notebook busca a série, recorta os
  anos fechados, confere contra o IBGE e formata.
- **Período publicado:** `1995-01-01` a `2025-01-01` (31 pontos).

## PIB per capita: nominal e real

- **`pib-per-capita-nominal`:** PIB per capita a preços correntes, em reais do próprio ano,
  de 1996 a 2025. 1996–2023 vêm da SIDRA 6784 (variável 9812) e 2024–2025 do caderno das
  Contas Nacionais Trimestrais (4º tri), que publica o nível corrente no fechamento anual.
  É o número oficial — o mesmo que o IBGE divulga como "PIB per capita de R$ 51.693,92 em
  2023" e R$ 59.687,49 em 2025.
- **`pib-per-capita-real`:** PIB per capita em reais constantes de **2023**, calculado pelo
  site a partir do nominal e da variação em volume oficial do per capita (9814) para trás,
  e da variação real do caderno das Contas Trimestrais para frente:
  `L_2023 = 9812_2023`, `L_t = L_{t+1} / (1 + 9814_{t+1}/100)` e
  `L_{t+1} = L_t × (1 + variação_real_{t+1}/100)`.

### Por que 2023

2023 é o último ano da tabela 6784, que ancora o nível oficial; 2024 e 2025 vêm do
fechamento anual das Contas Nacionais Trimestrais. Ancorar em 2023 faz o ponto de 2023 da
série real ser exatamente o número oficial, o que o leitor pode conferir. O ano-base é uma
escolha de apresentação — qualquer ano
serviria, desde que **fixo**. Em reais de 1995 os valores seriam pouco legíveis; em reais
de 2023 eles conversam com o número que está no noticiário. O ano-base não é rebaseado a
cada atualização: rebasear reescreveria todo o passado e misturaria revisão da fonte com
mudança de régua.

### Por que duas séries

O nominal sozinho engana: no recorte 1996–2023 ele cresceu **890,4%**, mas o real cresceu
**38,7%**. A diferença é o deflator implícito do PIB, que multiplicou os preços por 7,12
no período. As duas séries têm a mesma cobertura, a mesma unidade (`R$`) e a mesma
`medida` (`nivel`); o que muda é a régua — e a descrição de cada uma diz isso.

### Conferência do cálculo real

A rota principal encadeia a variação em volume oficial do PIB per capita. Duas rotas
independentes usam a mesma tabela 6784: o deflator implícito do PIB (9811) encadeado e o
PIB per capita a preços do ano anterior (9813) encadeado. As três têm de convergir dentro
de 0,5% em todo o período; o desvio medido é de 0,30% no pior ano (R$ 140 em 2008), efeito do
arredondamento da fonte.

### Cobertura

As duas séries começam em **1996**, primeiro ano das Contas Nacionais Anuais na tabela
6784, e terminam em **2025**. 1996–2023 vêm da 6784; 2024–2025 vêm do caderno das Contas
Nacionais Trimestrais (4º tri), que publica o nível per capita corrente e a variação real
no fechamento anual.

## Decisão metodológica (variação real anual)

O IBGE **publica** a taxa pronta em duas tabelas, ambas com **1 casa decimal**: a **6784**
(Contas Nacionais Anuais, variável 9810 — “PIB – variação em volume”), de 1996 a **2023**, e a
**5932** (Contas Nacionais Trimestrais, variável 6563 — “Taxa acumulada ao longo do ano”), cujo
4º trimestre é o ano fechado, de 1996 a **2025**.

A série publicada fica na **SGS 7326** por três motivos: (a) entrega a mesma taxa com **2
casas**, a precisão que a série já publica; (b) cobre até **2025**, enquanto a tabela anual para
em 2023; (c) a SGS 7326 é idêntica, ao centésimo, à taxa recomposta a partir do **índice de
volume encadeado** do IBGE (SIDRA 1620, “PIB a preços de mercado”, base média 1995 = 100) em
todos os anos de 1997 a 2025 — identidade conferida a cada execução, contra a taxa oficial
(tolerância de 0,05 p.p., o arredondamento de 1 casa) e contra o índice (0,01 p.p.).

A tabela anual 6784 **não** é usada como fonte da variação anual porque, nesta edição, ela
discorda das tabelas de nível em **2018 e 2019** (1,7 e 1,3 contra 1,8 e 1,2 em 1620/5932, e
1,78/1,22 na SGS). A mesma tabela 6784 **é** a fonte do PIB per capita, onde não há série
alternativa de nível; a divergência de 2018–2019 é da variação anual, não dos níveis per
capita. A execução imprime a divergência.

## Recorte de início e ano fechado

A série `pib-real-anual` começa em **1995**, alinhada às demais famílias do repositório, que
começam em `1995-01-01`. A SGS 7326 vai até 1963, mas os anos anteriores a 1995 vêm da série
antiga (base de volume anterior): a cadeia atual começa em 1995 — o índice do IBGE tem base
média 1995 = 100 e o primeiro trimestre publicado é 1996T1. Estender ganharia 32 pontos
(1963–1994) ao custo de misturar bases de volume, por isso o corte é mantido.

O ponto de **1995** é o único que as tabelas atuais do IBGE não permitem conferir: a cadeia
de volume começa no nível de 1995 e a primeira taxa publicada é a de 1996. Ele vem da SGS
7326 (4,22%) e é mantido para preservar o início de 1995 das demais séries; quem quiser
estritamente a base atual pode cortar a série em 1996.

O ano corrente nunca é publicado: em novembro de 2026 a fonte já fecha **2024** (3,42%) e
**2025** (2,29%), e 2026 ainda não tem taxa anual. As séries per capita usam a 6784 até
2023 e o caderno das Contas Trimestrais para 2024–2025; 2026 ainda não tem fechamento anual.

## Coleta da série oficial

A leitura é feita desde o início da série, sem atualização incremental e sem ler o JSON já
publicado para decidir o período. A SGS devolve uma leitura por ano, datada em `01/01`.


````python
from datetime import date
from decimal import Decimal, ROUND_HALF_UP
from pathlib import Path
import io
import json
import math
import os
import re
import time

import requests
try:
    from pypdf import PdfReader
except ImportError:  # Colab não traz pypdf; a versão é a mesma do pyproject.toml.
    import subprocess
    import sys

    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "pypdf==6.21.4"])
    from pypdf import PdfReader

RAIZ = Path(os.environ.get("DADOS_RAIZ", Path.cwd())).resolve()

def relativo(caminho):
    """Caminho exibido no notebook: relativo à raiz, nunca o absoluto da máquina."""
    try:
        return caminho.relative_to(RAIZ)
    except ValueError:
        return caminho.name
PASTA = RAIZ / "indicadores" / "pib" if (RAIZ / "indicadores" / "pib").is_dir() else RAIZ
DESTINO = Path(os.environ.get("PIB_SAIDA", PASTA))
ARQUIVO = DESTINO / "pib-real-anual.json"
ARQUIVO_PER_CAPITA_NOMINAL = DESTINO / "pib-per-capita-nominal.json"
ARQUIVO_PER_CAPITA_REAL = DESTINO / "pib-per-capita-real.json"

SERIE = 7326
URL_SGS = "https://api.bcb.gov.br/dados/serie/bcdata.sgs.{serie}/dados"
URL_FONTE = URL_SGS.format(serie=SERIE) + "?formato=json"
URL_6784 = "https://apisidra.ibge.gov.br/values/t/6784/n1/all/v/all/p/all?formato=json"
URL_SCNT = (
    "https://ftp.ibge.gov.br/Contas_Nacionais/Contas_Nacionais_Trimestrais/"
    "Fasciculo_Indicadores_IBGE/{ano}/pib-vol-val_{ano}04caderno.pdf"
)
PRIMEIRO_ANO = 1995
HOJE = date.today()


def baixar_json(url, params=None, tentativas=4):
    """GET com espera crescente; SGS e SIDRA falham de forma intermitente."""
    for tentativa in range(tentativas):
        try:
            resposta = requests.get(url, params=params, timeout=120)
            resposta.raise_for_status()
            return resposta.json()
        except (requests.RequestException, ValueError):
            if tentativa == tentativas - 1:
                raise
            time.sleep(2 ** tentativa)


def baixar_bytes(url, tentativas=4):
    """GET binário com espera crescente; devolve None quando o caderno não existe."""
    for tentativa in range(tentativas):
        try:
            resposta = requests.get(url, timeout=120)
            if resposta.status_code == 404:
                return None
            resposta.raise_for_status()
            return resposta.content
        except requests.RequestException:
            if tentativa == tentativas - 1:
                raise
            time.sleep(2 ** tentativa)


bruto = baixar_json(URL_SGS.format(serie=SERIE), {"formato": "json"})
if not bruto:
    raise ValueError(f"a série SGS {SERIE} não retornou dados")

anual = sorted(
    (date(int(item["data"][-4:]), int(item["data"][3:5]), int(item["data"][:2])), float(item["valor"]))
    for item in bruto
)

if len({dia for dia, _ in anual}) != len(anual):
    raise ValueError(f"a SGS {SERIE} devolveu anos repetidos")
if any((dia.day, dia.month) != (1, 1) for dia, _ in anual):
    raise ValueError(f"a SGS {SERIE} devolveu data fora de 01/01")
for (anterior, _), (seguinte, _) in zip(anual, anual[1:]):
    if seguinte.year - anterior.year != 1:
        raise ValueError(f"lacuna na SGS {SERIE}: {anterior} -> {seguinte}")

print(f"Série SGS {SERIE}: {len(anual)} leituras anuais")
print(f"Período bruto: {anual[0][0]:%Y-%m-%d} a {anual[-1][0]:%Y-%m-%d}")
print(f"Arquivos de saída: {relativo(ARQUIVO)}, "
      f"{relativo(ARQUIVO_PER_CAPITA_NOMINAL)} e {relativo(ARQUIVO_PER_CAPITA_REAL)}")
print("\nPrimeiras 3 leituras:")
for dia, valor in anual[:3]:
    print(f"  {dia:%Y-%m-%d}  {valor:7.2f}%")
print("Últimas 3 leituras:")
for dia, valor in anual[-3:]:
    print(f"  {dia:%Y-%m-%d}  {valor:7.2f}%")
````

## Recorte dos anos fechados e conferência contra o IBGE

O recorte começa em 1995 e exclui o ano corrente, que ainda não tem taxa anual. Depois, cada
ponto é conferido contra as duas publicações do IBGE: a taxa oficial das Contas Nacionais
Trimestrais (SIDRA 5932, variável 6563, 4º trimestre) e o índice de volume encadeado (SIDRA
1620, variável 583, “PIB a preços de mercado”), recompondo o ano pela razão entre a soma dos 4
trimestres do ano e a do ano anterior. Divergência acima da tolerância interrompe a publicação.

A tabela anual 6784 é lida apenas para registro: ela para em 2023 e, nesta edição, discorda das
tabelas de nível em 2018 e 2019.


````python
URL_SIDRA = (
    "https://apisidra.ibge.gov.br/values/t/{tabela}/n1/all/v/{variavel}/p/all"
    "/c11255/{categoria}?formato=json"
)
PIB_A_PRECOS_DE_MERCADO = "90707"
SEM_DADO = {"..", "...", "-", "X"}


def sidra_trimestral(tabela, variavel):
    """Leituras trimestrais do PIB a preços de mercado: {\"202504\": valor}."""
    linhas = baixar_json(
        URL_SIDRA.format(tabela=tabela, variavel=variavel, categoria=PIB_A_PRECOS_DE_MERCADO)
    )
    return {
        linha["D3C"]: float(linha["V"])
        for linha in linhas[1:]
        if linha["V"] not in SEM_DADO
    }


fechados = [
    (dia, valor) for dia, valor in anual if PRIMEIRO_ANO <= dia.year < HOJE.year
]
if not fechados:
    raise ValueError(f"nenhum ano fechado a partir de {PRIMEIRO_ANO}")
if fechados[-1][0].year < HOJE.year - 1:
    print(f"AVISO: a fonte não tem o ano {HOJE.year - 1}; último ano fechado é {fechados[-1][0].year}")

# Taxa oficial do IBGE: "Taxa acumulada ao longo do ano" (6563); no 4º trimestre é o ano civil.
taxa_ibge = {
    int(periodo[:4]): valor
    for periodo, valor in sidra_trimestral(5932, 6563).items()
    if periodo.endswith("4")
}

# Índice de volume encadeado (base média 1995 = 100), por ano e trimestre.
indice_ibge = {}
for periodo, valor in sidra_trimestral(1620, 583).items():
    indice_ibge.setdefault(int(periodo[:4]), {})[int(periodo[4:])] = valor

# A taxa anual é a razão entre a soma dos 4 trimestres do ano e a do ano anterior, menos 1.
composta_ibge = {}
for ano in sorted(indice_ibge):
    atual = indice_ibge[ano]
    anterior = indice_ibge.get(ano - 1, {})
    if len(atual) == 4 and len(anterior) == 4:
        composta_ibge[ano] = (sum(atual.values()) / sum(anterior.values()) - 1) * 100

TOLERANCIA_TAXA = 0.051    # meio décimo: o IBGE publica a taxa com 1 casa decimal
TOLERANCIA_INDICE = 0.01   # o índice é publicado com 2 casas decimais

# A série do IBGE começa em 1996: a cadeia de volume tem nível em 1995, mas não taxa.
SEM_CONFERENCIA_IBGE = {1995}

for dia, valor in fechados:
    ano = dia.year
    if ano in SEM_CONFERENCIA_IBGE:
        continue
    if ano not in taxa_ibge:
        raise ValueError(f"{ano} sem taxa no SIDRA 5932")
    if abs(valor - taxa_ibge[ano]) > TOLERANCIA_TAXA:
        raise ValueError(f"{ano}: SGS {valor} x taxa IBGE {taxa_ibge[ano]}")
    # O índice só permite recompor a partir de 1997 (falta 1995 para a base de 1996).
    if ano in composta_ibge and abs(valor - composta_ibge[ano]) > TOLERANCIA_INDICE:
        raise ValueError(f"{ano}: SGS {valor} x índice IBGE {composta_ibge[ano]:.4f}")

print(f"Anos publicados: {len(fechados)} ({fechados[0][0]:%Y} a {fechados[-1][0]:%Y})")
print(
    f"Conferidos contra a taxa do IBGE (SIDRA 5932/6563): "
    f"{len(fechados) - len(SEM_CONFERENCIA_IBGE)} anos; "
    f"1995 fica fora porque o IBGE só publica taxa a partir de 1996."
)
print(
    f"Conferidos contra o índice de volume (SIDRA 1620/583): "
    f"{len(fechados) - len(SEM_CONFERENCIA_IBGE) - 1} anos; "
    f"1996 fica fora porque o índice começa em 1996T1 e a base de 1996 é 1995."
)
print("\nAno  SGS 7326  taxa IBGE  índice IBGE")
for dia, valor in fechados:
    if dia.year not in (1995, 1996, 2003, 2009, 2015, 2016, 2020, 2021, 2024, 2025):
        continue
    taxa = taxa_ibge.get(dia.year)
    composta = composta_ibge.get(dia.year)
    texto_taxa = "—" if taxa is None else f"{taxa:.1f}"
    texto_indice = "—" if composta is None else f"{composta:.3f}"
    print(f"{dia.year}  {valor:8.2f}  {texto_taxa:>9}  {texto_indice:>11}")

# Tabela anual 6784: fonte dos níveis per capita e registro da divergência da variação anual.
linhas_anuais = baixar_json(URL_6784)
ANUAL_6784 = {}
for linha in linhas_anuais[1:]:
    if linha["V"] in SEM_DADO:
        continue
    ANUAL_6784.setdefault(int(linha["D2C"]), {})[int(linha["D3C"])] = float(linha["V"])
anual_ibge = ANUAL_6784[9810]
divergentes = [
    (dia.year, valor, anual_ibge[dia.year])
    for dia, valor in fechados
    if dia.year in anual_ibge and abs(valor - anual_ibge[dia.year]) > TOLERANCIA_TAXA
]
print(
    f"\nTabela 6784 (Contas Nacionais Anuais): "
    f"{min(anual_ibge)}–{max(anual_ibge)}, {len(anual_ibge)} anos."
)
if divergentes:
    detalhe = ", ".join(f"{ano} ({sgs:.2f} x {ibge:.1f})" for ano, sgs, ibge in divergentes)
    print(f"  divergências acima de {TOLERANCIA_TAXA} p.p.: {detalhe}")
    print("  não usada como fonte da variação anual: as tabelas de nível (1620) e a taxa trimestral (5932) concordam com a SGS.")
````

## PIB per capita — nominal e real (R$ de 2023)

O IBGE publica o PIB per capita a preços correntes (variável 9812) e a variação em volume
do PIB per capita (9814) de 1996 a 2023; para 2024–2025, o caderno das Contas Trimestrais
(4º tri) publica o nível corrente e a variação real do ano. A primeira série é o número
oficial em reais correntes; a segunda mede o crescimento real, mas não tem nível. Para
comparar poder de compra ao longo do tempo, o notebook ancora o nível real no valor oficial
de 2023, encadeia a variação em volume oficial para trás e a variação real do caderno para
frente: `L_2023 = 9812_2023`, `L_t = L_{t+1} / (1 + 9814_{t+1}/100)` e
`L_{t+1} = L_t × (1 + variação_real_{t+1}/100)`.

O ano-base fica **fixo** em 2023. Rebasear a cada atualização reescreveria todo o passado e
misturaria revisão da fonte com mudança de régua; uma mudança de base é decisão explícita,
não efeito colateral de rodar o notebook de novo.

A conferência usa duas rotas independentes da mesma tabela: o deflator implícito do PIB
(9811) encadeado e o PIB per capita a preços do ano anterior (9813) encadeado. As três
rotas têm de convergir dentro do arredondamento da fonte (tolerância de 0,5%).


````python
ANOS_6784 = sorted(ANUAL_6784[9812])
BASE_PER_CAPITA = 2023
assert BASE_PER_CAPITA in ANOS_6784, BASE_PER_CAPITA


def ler_per_capita_scnt(ano, conteudo):
    """Extrai do caderno das Contas Trimestrais o nível corrente e a variação real do ano."""
    leitor = PdfReader(io.BytesIO(conteudo))
    texto = " ".join((pagina.extract_text() or "") for pagina in leitor.pages)
    texto = " ".join(texto.split())
    padrao = re.search(
        r"PIB per capita alcançou R\$ ([\d.]+,\d{2}).*?avanço.*?de ([\d,]+)%",
        texto,
        re.IGNORECASE,
    )
    if not padrao:
        raise ValueError(f"{ano}: não encontrei o PIB per capita no caderno do IBGE")
    nivel = float(padrao.group(1).replace(".", "").replace(",", "."))
    variacao = float(padrao.group(2).replace(",", "."))
    return nivel, variacao


# 2024 e 2025 não estão na tabela anual 6784: o nível corrente e a variação real do
# ano vêm do caderno das Contas Nacionais Trimestrais (4º tri), que fecha o ano.
SCNT = {}
for ano in range(max(ANOS_6784) + 1, HOJE.year):
    conteudo = baixar_bytes(URL_SCNT.format(ano=ano))
    if conteudo is None:
        raise ValueError(f"{ano}: caderno das Contas Trimestrais não encontrado")
    nivel, variacao = ler_per_capita_scnt(ano, conteudo)
    SCNT[ano] = {"nivel": nivel, "variacao": variacao}

ANOS_PER_CAPITA = sorted(set(ANOS_6784) | set(SCNT))
nominal = {ano: ANUAL_6784[9812][ano] for ano in ANOS_6784}
for ano, dados in SCNT.items():
    nominal[ano] = dados["nivel"]

# Rota principal: nível real ancorado no nominal do ano-base, encadeando a variação em
# volume oficial do PIB per capita (9814) para trás e a variação real do caderno para
# frente, nos anos que a 6784 ainda não publica.
real = {BASE_PER_CAPITA: nominal[BASE_PER_CAPITA]}
for ano in range(BASE_PER_CAPITA - 1, ANOS_6784[0] - 1, -1):
    real[ano] = real[ano + 1] / (1 + ANUAL_6784[9814][ano + 1] / 100)
for ano in sorted(SCNT):
    real[ano] = real[ano - 1] * (1 + SCNT[ano]["variacao"] / 100)

# Rota B: nominal deflacionado pelo deflator implícito do PIB (9811) encadeado.
deflator = {ANOS_6784[0]: 1.0}
for ano in ANOS_6784[1:]:
    deflator[ano] = deflator[ano - 1] * (1 + ANUAL_6784[9811][ano] / 100)
real_deflator = {
    ano: nominal[ano] * (deflator[BASE_PER_CAPITA] / deflator[ano])
    for ano in ANOS_6784
}

# Rota C: PIB per capita a preços do ano anterior (9813) encadeado.
real_ano_anterior = {BASE_PER_CAPITA: nominal[BASE_PER_CAPITA]}
for ano in range(BASE_PER_CAPITA - 1, ANOS_6784[0] - 1, -1):
    real_ano_anterior[ano] = (
        real_ano_anterior[ano + 1] * nominal[ano] / ANUAL_6784[9813][ano + 1]
    )

TOLERANCIA_PER_CAPITA = 0.005  # 0,5%: medido 0,30% no pior ano (arredondamento da fonte)
for ano in ANOS_6784:
    for nome, valor in (
        ("deflator", real_deflator[ano]),
        ("preços do ano anterior", real_ano_anterior[ano]),
    ):
        desvio = abs(real[ano] - valor) / real[ano]
        assert desvio <= TOLERANCIA_PER_CAPITA, (ano, nome, real[ano], valor, desvio)

for ano, dados in SCNT.items():
    assert nominal[ano] == dados["nivel"], (ano, nominal[ano], dados["nivel"])
    variacao = (real[ano] / real[ano - 1] - 1) * 100
    assert abs(variacao - dados["variacao"]) < 1e-9, (ano, variacao, dados["variacao"])

crescimento_real = (real[ANOS_PER_CAPITA[-1]] / real[ANOS_PER_CAPITA[0]] - 1) * 100
crescimento_nominal = (nominal[ANOS_PER_CAPITA[-1]] / nominal[ANOS_PER_CAPITA[0]] - 1) * 100
print(f"PIB per capita: {len(ANOS_PER_CAPITA)} anos ({ANOS_PER_CAPITA[0]}–{ANOS_PER_CAPITA[-1]})")
print(f"Nominal: R$ {nominal[ANOS_PER_CAPITA[0]]:.2f} → R$ {nominal[ANOS_PER_CAPITA[-1]]:.2f} "
      f"({crescimento_nominal:.1f}%)")
print(f"Real (R$ de {BASE_PER_CAPITA}): R$ {real[ANOS_PER_CAPITA[0]]:.2f} → "
      f"R$ {real[ANOS_PER_CAPITA[-1]]:.2f} ({crescimento_real:.1f}%)")
print(f"2024–2025 lidos do caderno das Contas Trimestrais: {sorted(SCNT)}")
````

## Escrita dos JSONs do contrato

Cada arquivo é gravado inteiro (nada de merge com o que já está publicado), com `versao`,
`id`, `unidade`, `eixo`, `medida`, `granularidade`, `fonte`, `urlFonte` e `dados` no formato
`{"data": "YYYY-MM-DD", "valor": ...}`. São três arquivos: `pib-real-anual.json`,
`pib-per-capita-nominal.json` e `pib-per-capita-real.json`.


````python
def arredondar(valor):
    """Arredonda para 2 casas com meio para cima (convenção comercial)."""
    return float(Decimal(repr(valor)).quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))


pontos = [
    {"data": dia.strftime("%Y-%m-%d"), "valor": arredondar(valor)} for dia, valor in fechados
]

serie = {
    "versao": 1,
    "id": "pib-real-anual",
    "unidade": {"tipo": "percentual", "simbolo": "%", "casas": 2},
    "eixo": "percentual",
    "medida": "variacao-anual",
    "granularidade": "anual",
    "fonte": "IBGE",
    "urlFonte": URL_FONTE,
    "dados": pontos,
}


def montar_per_capita(id_, valores, unidade, fonte):
    return {
        "versao": 1,
        "id": id_,
        "unidade": unidade,
        "eixo": "reais",
        "medida": "nivel",
        "granularidade": "anual",
        "fonte": fonte,
        "urlFonte": URL_6784,
        "dados": [
            {"data": f"{ano}-01-01", "valor": arredondar(valores[ano])}
            for ano in ANOS_PER_CAPITA
        ],
    }


serie_per_capita_nominal = montar_per_capita(
    "pib-per-capita-nominal",
    nominal,
    {"tipo": "moeda", "simbolo": "R$", "casas": 2},
    "IBGE (Contas Nacionais Anuais, SIDRA 6784, até 2023; Contas Nacionais Trimestrais, 4º tri, para 2024–2025)",
)
serie_per_capita_real = montar_per_capita(
    "pib-per-capita-real",
    real,
    {"tipo": "moeda", "simbolo": "R$", "casas": 2},
    "IBGE (Contas Nacionais Anuais, SIDRA 6784, até 2023; Contas Nacionais Trimestrais, 4º tri, para 2024–2025); cálculo do site (R$ de 2023)",
)

DESTINO.mkdir(parents=True, exist_ok=True)
for caminho, objeto in (
    (ARQUIVO, serie),
    (ARQUIVO_PER_CAPITA_NOMINAL, serie_per_capita_nominal),
    (ARQUIVO_PER_CAPITA_REAL, serie_per_capita_real),
):
    with caminho.open("w", encoding="utf-8") as arquivo:
        json.dump(objeto, arquivo, ensure_ascii=False, indent=2)
        arquivo.write("\n")
    print(f"{len(objeto['dados'])} pontos escritos em {relativo(caminho)}")
````

## Validação inline

A célula final falha se o contrato, a ordem, a unicidade dos anos, a finitude dos valores, a
igualdade com a fonte, a conferência contra o IBGE ou a coerência entre o per capita nominal
e o real forem violados. Também confere os números citados em `metodologia.md`.


````python
CHAVES_RAIZ = {
    "versao", "id", "unidade", "eixo", "medida", "granularidade", "fonte", "urlFonte", "dados",
}
assert set(serie) == CHAVES_RAIZ, sorted(set(serie) ^ CHAVES_RAIZ)
assert serie["versao"] == 1, serie["versao"]
assert serie["id"] == "pib-real-anual", serie["id"]
assert serie["unidade"] == {"tipo": "percentual", "simbolo": "%", "casas": 2}, serie["unidade"]
assert serie["eixo"] == "percentual", serie["eixo"]
assert serie["medida"] == "variacao-anual", serie["medida"]
assert serie["granularidade"] == "anual", serie["granularidade"]
assert serie["fonte"] == "IBGE", serie["fonte"]
assert serie["urlFonte"].startswith("https://"), serie["urlFonte"]

dados = serie["dados"]
assert dados, "série vazia"

datas = [ponto["data"] for ponto in dados]
dias = [date.fromisoformat(texto) for texto in datas]
por_ano = {ponto["data"][:4]: ponto["valor"] for ponto in dados}

for ponto in dados:
    assert set(ponto) == {"data", "valor"}, sorted(ponto)
    assert isinstance(ponto["data"], str), ponto
    assert not isinstance(ponto["valor"], bool), ponto
    assert isinstance(ponto["valor"], (int, float)), ponto
    assert math.isfinite(ponto["valor"]), ponto

assert all(a < b for a, b in zip(dias, dias[1:])), "datas fora de ordem ou repetidas"
assert all((dia.day, dia.month) == (1, 1) for dia in dias), "data fora de 01/01"
assert all(b.year - a.year == 1 for a, b in zip(dias, dias[1:])), "ano faltando ou repetido"
assert datas[0] == "1995-01-01", datas[0]
assert datas[-1] == f"{fechados[-1][0].year}-01-01", datas[-1]
assert int(datas[-1][:4]) >= HOJE.year - 1, datas[-1]
assert len(dados) == len(fechados), (len(dados), len(fechados))

# 1) O valor publicado é a leitura da fonte, arredondada a 2 casas.
esperado = [
    {"data": dia.strftime("%Y-%m-%d"), "valor": arredondar(valor)} for dia, valor in fechados
]
assert dados == esperado, "valor publicado difere da fonte"

# 2) Cada ponto continua coerente com o IBGE (taxa oficial e índice de volume).
for dia, valor in fechados:
    if dia.year in SEM_CONFERENCIA_IBGE:
        continue
    assert abs(valor - taxa_ibge[dia.year]) <= TOLERANCIA_TAXA, (dia.year, valor, taxa_ibge[dia.year])
    if dia.year in composta_ibge:
        assert abs(valor - composta_ibge[dia.year]) <= TOLERANCIA_INDICE, (dia.year, valor)

# 3) Números citados em metodologia.md (esta edição da fonte).
citados = {
    "1995": 4.22,
    "2003": 1.14,
    "2009": -0.13,
    "2015": -3.55,
    "2016": -3.28,
    "2020": -3.28,
    "2021": 4.76,
    "2024": 3.42,
    "2025": 2.29,
}
for ano, valor in citados.items():
    assert por_ano[ano] == valor, (ano, por_ano[ano], valor)

# 4) O PIB de 2020 encolheu e o de 2021 recuperou: o sinal tem de refletir o choque.
assert por_ano["2020"] < 0 < por_ano["2021"], (por_ano["2020"], por_ano["2021"])

with ARQUIVO.open(encoding="utf-8") as arquivo:
    salvo = json.load(arquivo)
assert salvo == serie, "arquivo gravado difere do objeto validado"


# ---------------------------------------------------------------- per capita
UNIDADE_PER_CAPITA = {"tipo": "moeda", "simbolo": "R$", "casas": 2}
FONTE_NOMINAL = (
    "IBGE (Contas Nacionais Anuais, SIDRA 6784, até 2023; "
    "Contas Nacionais Trimestrais, 4º tri, para 2024–2025)"
)
FONTE_REAL = (
    "IBGE (Contas Nacionais Anuais, SIDRA 6784, até 2023; "
    "Contas Nacionais Trimestrais, 4º tri, para 2024–2025); cálculo do site (R$ de 2023)"
)


def validar_per_capita(objeto, id_, valores, fonte, caminho):
    assert set(objeto) == CHAVES_RAIZ, sorted(set(objeto) ^ CHAVES_RAIZ)
    assert objeto["versao"] == 1, objeto["versao"]
    assert objeto["id"] == id_, objeto["id"]
    assert objeto["unidade"] == UNIDADE_PER_CAPITA, objeto["unidade"]
    assert objeto["eixo"] == "reais", objeto["eixo"]
    assert objeto["medida"] == "nivel", objeto["medida"]
    assert objeto["granularidade"] == "anual", objeto["granularidade"]
    assert objeto["fonte"] == fonte, objeto["fonte"]
    assert objeto["urlFonte"] == URL_6784, objeto["urlFonte"]

    pontos_ = objeto["dados"]
    assert len(pontos_) == len(ANOS_PER_CAPITA), (len(pontos_), len(ANOS_PER_CAPITA))
    assert [p["data"] for p in pontos_] == [f"{ano}-01-01" for ano in ANOS_PER_CAPITA]
    assert pontos_[0]["data"] == "1996-01-01", pontos_[0]
    assert pontos_[-1]["data"] == f"{ANOS_PER_CAPITA[-1]}-01-01", pontos_[-1]
    for ponto in pontos_:
        assert set(ponto) == {"data", "valor"}, sorted(ponto)
        assert isinstance(ponto["data"], str), ponto
        assert not isinstance(ponto["valor"], bool), ponto
        assert isinstance(ponto["valor"], (int, float)), ponto
        assert math.isfinite(ponto["valor"]), ponto
    esperado_ = [
        {"data": f"{ano}-01-01", "valor": arredondar(valores[ano])}
        for ano in ANOS_PER_CAPITA
    ]
    assert pontos_ == esperado_, f"{id_}: valor publicado difere do cálculo"
    with caminho.open(encoding="utf-8") as arquivo:
        assert json.load(arquivo) == objeto, f"{id_}: arquivo gravado difere do objeto"


validar_per_capita(serie_per_capita_nominal, "pib-per-capita-nominal", nominal, FONTE_NOMINAL, ARQUIVO_PER_CAPITA_NOMINAL)
validar_per_capita(serie_per_capita_real, "pib-per-capita-real", real, FONTE_REAL, ARQUIVO_PER_CAPITA_REAL)

pc_nominal = {p["data"][:4]: p["valor"] for p in serie_per_capita_nominal["dados"]}
pc_real = {p["data"][:4]: p["valor"] for p in serie_per_capita_real["dados"]}

# 5) As duas séries contam a mesma história com réguas diferentes.
assert list(pc_nominal) == list(pc_real), "coberturas diferentes entre nominal e real"
assert pc_real["2023"] == pc_nominal["2023"], "o ponto do ano-base tem de coincidir"
assert pc_real["1996"] > pc_nominal["1996"], "em 1996 o real vale mais que o nominal"
assert pc_nominal["1996"] == 5219.36, pc_nominal["1996"]
assert pc_nominal["2023"] == 51693.92, pc_nominal["2023"]
assert pc_nominal["2024"] == 55247.45, pc_nominal["2024"]
assert pc_nominal["2025"] == 59687.49, pc_nominal["2025"]
assert pc_real["1996"] == 37261.36, pc_real["1996"]
assert pc_real["2023"] == 51693.92, pc_real["2023"]
assert round((pc_real["2024"] / pc_real["2023"] - 1) * 100, 1) == 3.0
assert round((pc_real["2025"] / pc_real["2024"] - 1) * 100, 1) == 1.9
assert round((pc_real["2023"] / pc_real["1996"] - 1) * 100, 1) == 38.7
assert round((pc_nominal["2023"] / pc_nominal["1996"] - 1) * 100, 1) == 890.4
assert (pc_nominal["2023"] / pc_nominal["1996"]) > (pc_real["2023"] / pc_real["1996"])

print(
    f"ok: {len(dados)} anos únicos, {datas[0]} a {datas[-1]}, "
    f"leituras idênticas à SGS {SERIE} e coerentes com o SIDRA (5932 e 1620); "
    f"per capita nominal e real com cobertura idêntica ({ANOS_PER_CAPITA[0]}–{BASE_PER_CAPITA})"
)
````
